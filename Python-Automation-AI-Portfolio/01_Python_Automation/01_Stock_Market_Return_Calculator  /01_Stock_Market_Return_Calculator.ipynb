```python
amount = float(input("Enter your initial investent (Euro):"))
rate =float(input("Enter your annual return (%):"))
years = int(input("Enter the investment period (years):"))

for year in range (1, years+1):
  amount = amount * (1+ rate/100)
  print(f"Year {year}: {amount:.2f} Euro")



```
