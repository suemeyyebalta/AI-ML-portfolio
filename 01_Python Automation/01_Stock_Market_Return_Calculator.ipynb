```python
amountStr = input("Bitte geben Sie Ihre Ersparnisse ein: ")
amount = float(amountStr)

amount = amount * 1.08
print("Nach 1 Jahr haben Sie: " + str(amount) + " Euro")

for i in range(2, 41):
    amount = amount * 1.08
    print("Nach " + str(i) + " Jahren haben Sie: " + str(amount) + " Euro")
```
