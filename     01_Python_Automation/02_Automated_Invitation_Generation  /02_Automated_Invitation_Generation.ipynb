```python
names = [
    'James Wilson',
    'Oliver Brown'
]

for name in names:
    print("------------------------")
    print(f"Invitation to: {name}")
    print("------------------------")
    text = f"Dear {name}, \n\n" + \
        "We would like to invite you to our annual company event.\n\n" + \
        "Best regards,\n" + \
        "Your Company"

    print(text)
```

```python
with open("guests.txt", "r", encoding="utf-8") as file:
    for name in file:
        name = name.strip()
        print("------------------------")
        print(f"Invitation to: {name}")
        print("------------------------")
        text = f"Dear {name}, \n\n" + \
        "We would like to invite you to our annual company event.\n\n" + \
        "Best regards,\n" + \
        "Your Company"

        print(text)
```

```python
import os
os.mkdir("invitations")
with open("guests.txt", "r", encoding="utf-8") as file:
    for name in file:
        name = name.strip()
        print("------------------------")
        print(f"Invitation to: {name}")
        print("------------------------")

        text = f"Dear {name}, \n\n" + \
        "We would like to invite you to our annual company event.\n\n" + \
        "Best regards,\n" + \
        "Your Company"

        pathToNewFile = os.path.join("invitations", f"{name}.txt")
        with open(pathToNewFile, 'w', encoding="utf-8") as outFile:
            outFile.write(text)

        print(text)
```
