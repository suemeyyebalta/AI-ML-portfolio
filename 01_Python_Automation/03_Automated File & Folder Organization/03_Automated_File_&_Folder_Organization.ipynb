```python
import os
import shutil

folder = "/content/Folder 1/files"
imageFolder = os.path.join(folder, "Images")
if not os.path.isdir(imageFolder):
    os.mkdir(imageFolder)

for filename in os.listdir(folder):
    if filename.endswith('.jpg') or filename.endswith('.JPG'):
        fileNameWithPath = os.path.join(folder, filename)
        outFileNameWithPath = os.path.join(imageFolder, filename)
        shutil.move(fileNameWithPath, outFileNameWithPath)
        print(fileNameWithPath)
        print(outFileNameWithPath)
        print("---------")
```

```python
import os
import shutil
folder = "/content/Folder 2/files"
imageFolder = os.path.join(folder, "Images")
if not os.path.isdir(imageFolder):
    os.mkdir(imageFolder)
for filename in os.listdir(folder):
    if filename.lower().endswith('.jpg'):
        fileNameWithPath = os.path.join(folder, filename)
        outFileNameWithPath = os.path.join(imageFolder, filename)
        shutil.move(fileNameWithPath, outFileNameWithPath)
        print(fileNameWithPath)
        print(outFileNameWithPath)
        print("---------")
```

```python
import os
import shutil

baseFolder = "/content"

for foldername in os.listdir(baseFolder):

    folder = os.path.join(baseFolder, foldername, "files")

    if not os.path.isdir(folder):
        continue

    imageFolder = os.path.join(folder, "Images")
    documentFolder = os.path.join(folder, "Documents")
    otherFolder = os.path.join(folder, "Others")

    if not os.path.isdir(imageFolder):
        os.mkdir(imageFolder)

    if not os.path.isdir(documentFolder):
        os.mkdir(documentFolder)

    if not os.path.isdir(otherFolder):
        os.mkdir(otherFolder)

    for filename in os.listdir(folder):

        fileNameWithPath = os.path.join(folder, filename)

        if os.path.isdir(fileNameWithPath):
            continue

        if filename.lower().endswith(".jpg"):
            outFileNameWithPath = os.path.join(imageFolder, filename)

        elif filename.lower().endswith((".txt", ".docx")):
            outFileNameWithPath = os.path.join(documentFolder, filename)

        else:
            outFileNameWithPath = os.path.join(otherFolder, filename)

        shutil.move(fileNameWithPath, outFileNameWithPath)

        print(fileNameWithPath)
        print(outFileNameWithPath)
        print("--------------------")
```
